# Inference Load Test Planning Starter

Plan load test scenarios before productionizing an LLM inference service.

```python
import pandas as pd

scenarios = pd.DataFrame(
    {
        'scenario': ['baseline', 'peak_traffic', 'long_context'],
        'rps_target': [5, 25, 10],
        'p95_latency_target_ms': [1500, 2500, 4000],
        'error_rate_target': [0.01, 0.03, 0.02],
    }
)
scenarios
```

```python
def evaluate_slo(actual_p95, actual_err, p95_target, err_target):
    return actual_p95 <= p95_target and actual_err <= err_target

print('Example SLO pass:', evaluate_slo(1400, 0.01, 1500, 0.02))
```
