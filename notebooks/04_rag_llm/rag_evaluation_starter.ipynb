# RAG Evaluation Starter

Starter notebook for evaluating retrieval and answer quality in a RAG system.

```python
import pandas as pd

samples = pd.DataFrame(
    {
        'question': ['What is feature store?', 'What is idempotency?'],
        'retrieved_context_relevant': [1, 0],
        'answer_faithful': [1, 0],
        'answer_helpful': [1, 1],
    }
)
samples
```

```python
metrics = {
    'context_relevance_rate': samples['retrieved_context_relevant'].mean(),
    'faithfulness_rate': samples['answer_faithful'].mean(),
    'helpfulness_rate': samples['answer_helpful'].mean(),
}
metrics
```
