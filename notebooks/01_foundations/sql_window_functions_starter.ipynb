# SQL Window Functions Starter

Use this notebook to practice ranking and running totals quickly.

```python
import pandas as pd

df = pd.DataFrame(
    {
        'team': ['A', 'A', 'A', 'B', 'B'],
        'day': [1, 2, 3, 1, 2],
        'score': [10, 15, 8, 12, 20],
    }
)
df
```

```python
# Simulate SQL window operations using pandas for intuition.
df['running_sum'] = df.sort_values('day').groupby('team')['score'].cumsum()
df['rank_desc'] = df.groupby('team')['score'].rank(method='dense', ascending=False)
df.sort_values(['team', 'day'])
```
