# Embeddings Similarity Starter

Simple cosine similarity intuition with small vectors.

```python
import numpy as np

docs = {
    'doc_a': np.array([0.8, 0.1, 0.2]),
    'doc_b': np.array([0.75, 0.15, 0.1]),
    'doc_c': np.array([0.1, 0.9, 0.2]),
}
query = np.array([0.78, 0.12, 0.18])
```

```python
def cosine(a, b):
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))

scores = {k: cosine(query, v) for k, v in docs.items()}
sorted(scores.items(), key=lambda x: x[1], reverse=True)
```
