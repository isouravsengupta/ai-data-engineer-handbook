# Python List and Slicing Drills

Beginner-friendly notebook for mastering core list operations.

```python
nums = [1, 2, 3, 4, 5, 6]
print('Original:', nums)
print('Reverse:', nums[::-1])
print('Every second:', nums[::2])
print('Middle slice [1:4]:', nums[1:4])
```

```python
def rotate_right(values, k):
    k = k % len(values)
    return values[-k:] + values[:-k]

assert rotate_right([1, 2, 3, 4, 5], 2) == [4, 5, 1, 2, 3]
print('Rotate right test passed')
```
