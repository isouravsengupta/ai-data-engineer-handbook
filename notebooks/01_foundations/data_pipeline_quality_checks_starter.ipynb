# Data Pipeline Quality Checks Starter

Practice basic data quality checks that every pipeline should have.

```python
import pandas as pd

df = pd.DataFrame(
    {
        'id': [1, 2, 2, 4],
        'email': ['a@x.com', 'b@x.com', None, 'd@x.com'],
        'amount': [100, 200, -10, 300],
    }
)
df
```

```python
checks = {
    'null_email_count': int(df['email'].isna().sum()),
    'duplicate_id_count': int(df['id'].duplicated().sum()),
    'negative_amount_count': int((df['amount'] < 0).sum()),
}
checks
```
